```python
from openai import OpenAI

client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.environ["HF_ACCESS_TOKEN"],
)
```

Describe the image in a sentence:

```python
response = client.chat.completions.create(
    model="zai-org/GLM-5.3-Flash:together",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "Describe this image in one sentence."
                },
                {
                    "type": "image_url",
                    "image_url": {
                        "url": "https://cdn.britannica.com/61/93061-050-99147DCE/Statue-of-Liberty-Island-New-York-Bay.jpg"
                    }
                }
            ]
        }
    ],
)

print(response.choices[0].message.content)
```

Best Movie with the most views and positive reviews:

```python
response = client.chat.completions.create(
    model="zai-org/GLM-5.3:fireworks-ai",
    messages=[
        {
            "role": "user",
            "content": "What is the best movie in terms of most views and positive reviews? Explain the plot of movie in few sentences. Send response in Markdown format"
        }
    ],
)

print(response.choices[0].message.content)
```

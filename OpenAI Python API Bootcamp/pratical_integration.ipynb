### _Prompt construction:_
Best principles for effective prompts:
- Model choice - preferably latest models
- Instructions - put instructions at the start of prompt and seperate it from desired output using delimiters like ''' or ### 
- Provide details for specific output
- Examples - add desired format example after instructions to help model format response
- Direct - more direct in quantitative description (eg: create 3 to 5 sentences on a topic)
- Initiate the response - useful for code completion (eg: python function below to scrape wikipedia to return CSV of world's tallest buildings and their stats    import )
- Prevent hallucinations - tell the model in the prompt to only answer if information on a topic/company is found/100% true, otherwise return 'I don't know'.

Generate a Multiple choice Exam:

```python
def create_test_prompt(topic, num_questions, num_possible_answers):
    prompt = f"Create a multiple choice quiz on the topic of {topic} consisting of {num_questions} questions. Each question should have {num_possible_answers} options. Also include the correct answer for each question using the starting string 'Correct Answer:' "
    return prompt
```

### _Accept user input and send to the Model:_

```python
topic = input("On what topic would you like to create a multiple choice test? ")
```

```python
num_questions = int(input("How many number of questions do you want in the test?"))
```

```python
num_choices = int(input("How many number of options do you want for each question?"))
```

```python
prompt = create_test_prompt(topic,num_questions,num_choices)
print(prompt)
```

```python
from openai import OpenAI

client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.environ["HF_ACCESS_TOKEN"],
)
```

```python
response = client.chat.completions.create(
    model="zai-org/GLM-5.3:fireworks-ai",
    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0.7,
)

print(response.choices[0].message.content)
```

### _Process the response:_

```python
# Removing print statement shows newline \n in response that can be used to split response 
response.choices[0].message.content
```

```python
# Function to create student view of the multiple choice test
def create_student_view(test, num_questions):
    # Create dict to process questions - key 1 is an empty string
    student_view = {1:''}
    question_num = 1
    # Split each line using \n
    for line in test.split("\n"):
        if not line.startswith("Correct Answer:"):
            student_view[question_num] += line + '\n'
        else: 
            if question_num < num_questions:
                question_num += 1
                # Add new key with empty string for next question number
                student_view[question_num] = ''
    return student_view
```

```python
view_questions = create_student_view(response.choices[0].message.content, 4)
print(view_questions)
```

```python
for key in view_questions:
    print(view_questions[key])
```

```python
# Function to get correct answers of the multiple choice test
def get_answers(test, num_questions):
    # Create dict to get answers - key 1 is an empty string
    answers = {1:''}
    question_num = 1
    # Split each line using \n
    for line in test.split("\n"):
        if line.startswith("Correct Answer:"):
            answers[question_num] += line + '\n'

            if question_num < num_questions:
                question_num += 1
                # Add new key with empty string for next question number
                answers[question_num] = ''
    return answers
```

```python
correct_answers = get_answers(response.choices[0].message.content, 4)
print(correct_answers)
```

```python
# Student Exam Simulation
def exam(view_questions):
    # Store student's exam answers
    student_answers = {}
    for question_num, question in view_questions.items():
        print(question)
        answer = input(f"Enter your answer for Question {question_num}: ")
        student_answers[question_num] = answer
    return student_answers 
```

```python
student_answers = exam(view_questions)
```

```python
# Compare student answers with the correct answers using same question number key
def grade(correct_answers_dict, student_answers_dict):
    correct_answers = 0
    for question, answer in student_answers_dict.items():
        if answer.upper() == correct_answers_dict[question][16]:
            correct_answers += 1
    grade = 100*correct_answers/len(correct_answers_dict)

    return f"{correct_answers}/{len(correct_answers_dict)} correct!. Grade = {grade}%"
```

```python
grade(correct_answers, student_answers)
```

### _Reusable API function with basic error handling:_

Catch Exception OpenAI library raises:
- RateLimitError - Hit rate limit. Try again after some time
- APITimeoutError - The request timed out. Safe to retry.
- APIConnectionError - Network issue between your machine and the API.
- APIStatusError - Catch-all for other HTTP errors. Check the status code before retrying.
- AuthenticationError - Bad API key that needs to be fixed.
- BadRequestError - Request was malformed (invalid parameter, context too long etc). Fix the request.

```python
from openai import (
    OpenAI,
    RateLimitError,
    APITimeoutError,
    APIConnectionError,
    AuthenticationError,
    BadRequestError,
    APIStatusError,
)

client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.environ["HF_ACCESS_TOKEN_NAME"],
)

# Try and except for basic error handling in python
try:
    response = client.chat.completions.create(
        model="zai-org/GLM-5.3:fireworks-ai",
        messages=[
            {
                "role": "system",
                "content": "Insert system prompt instructions"
            },
            {
                "role": "user",
                "content": "Insert user text with user inputs"
            }            
        ],
    )
    print(response.choices[0].message.content)

# Run if specific error occurs
except AuthenticationError:
    print("Error: Invalid API key or unauthorized access. Fix the API key.")

except BadRequestError as e:
    print(f"Error: Request was invalid: {e}. Check the request text and model parameters.")

except RateLimitError:
    print("Error: You have hit your rate limit. Please wait before retrying.")

except APITimeoutError:
    print("Error: The request timed out.")

except APIConnectionError as e:
    print(f"Error: Failed to connect to the LLM service: {e}")

except APIStatusError as e:
    print(f"Error: API returned status {e.status_code}: {e.message}")
```
